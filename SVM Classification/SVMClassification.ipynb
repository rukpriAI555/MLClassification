```python
import pandas as pd
```

```python
dataset = pd.read_csv("Social_Network_Ads.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, dtype=int, drop_first = True)
```

```python
#Converting categorical data to numerical data and removing redundant column
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['User ID', 'Age', 'EstimatedSalary', 
       'Gender_Male']]
```

```python
independent
```

```python
dependent = dataset[["Purchased"]]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
#Splitting training dataset and test dataset above
```

```python
x_train
```

```python
#Using Standardisation technique to find the best model; x_train above - before standardisation
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
x_train=sc.fit_transform(x_train)
```

```python
x_test = sc.transform(x_test)
```

```python
#x_train
```

```python
#x_Train after standardisation technique above
```

```python
from sklearn.svm import SVC
```

```python
Classifier = SVC(kernel ='rbf',random_state = 0)
```

```python
Classifier.fit(x_train, y_train)
```

```python
#Classifier.intercept_
```

```python
#Viewing the origin or intercept
```

```python
#Classifier.n_support_
```

```python
#Viewing the number of support vectors used
```

```python
#Classifier.support_
```

```python
#Viewing the support vectors
```

```python
y_pred=Classifier.predict(x_test)
```

```python
#Evaluation Metrics for Support vector Classification
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm= confusion_matrix(y_test, y_pred)
```

```python
cm
```

```python
from sklearn.metrics import classification_report
```

```python
clf_report = classification_report(y_test, y_pred)
```
